# Data Visualization

```python
from matplotlib import pyplot as plt
```

## Line Chart

```python
years = [1950, 1960, 1970, 1980, 1990, 2000, 2010]
gdp = [300.2, 543.3, 1075.9, 2862.5, 5979.6, 10289.7, 14958.3]

plt.plot(
    years,
    gdp,
    color='green',
    marker='o',
    linestyle='solid'
)
plt.title('Nominal DGP')

plt.ylabel('Billions of $')
plt.show()
```

## Bar Chart

```python
movies = ['Annie Hall','Ben-Hur', 'Casablanca', 'Gandhi', 'West Side Story']
num_oscars = [5, 11, 3, 8, 10]
movies_range = range(len(movies))
plt.bar(movies_range, num_oscars)

plt.title("My favorite Movies")
plt.ylabel('# of Academy Awards')

# rotule o eixo x com o nome dos filmes no centro das barras
plt.xticks(movies_range, movies)

plt.show()
```

# Histogram

```python
from collections import Counter

grades = [83, 95, 91, 87, 70, 0, 85, 82, 100, 67, 73, 77, 0]

#Agrupe notas por decil, mas coloque o 100 com o 90

# '//' indica divisão inteira
histogram = Counter(min(grade // 10 * 10, 90) for grade in grades) # Counter([9, 1, 3]) -> Counter({9: 1, 1: 1, 3: 1})

plt.bar(
    [x + 5 for x in histogram.keys()], # Move as barras para a direita em 5
    histogram.values(), # Dá para o tamanho das barras
    10, #Largura das barras
    edgecolor=(0, 0, 0) # Cor da borda das barras
)
plt.axis([-5, 105, 0, 5]) # eixos x de -5 a 105, eixo y de 0 a 5

plt.xticks([10 * i for i in range(11)]) # rótulos do eixo x: 0, 10, 20, ..., 100

plt.xlabel('Decile')
plt.ylabel('# of students')
plt.title('Distribution of Exam 1 Grades')
plt.show()
```

## Line Chart

```python
variance = [2**i for i in range(9)]
bias_squared = variance[::-1]
total_error = [x + y for x, y in zip(variance, bias_squared)]
xs = [i for i, _ in enumerate(variance)]

# multiplas chamadas para o pyplot
# para mostrar multiplas séries no mesmo gráfico
plt.plot(xs, variance, 'g-', label='variance') # linha verde sólida
plt.plot(xs, bias_squared, 'r-.', label='bias squared') # linha vermelha com traços e pontos

plt.plot(xs, total_error, 'b:', label='total error') # linha azul pontilhada

# legenda
plt.legend(loc=9)

plt.xlabel('model complexity')
plt.xticks([]) # remove os ticks
plt.ylabel('The Bias-Variance Tradeoff')

plt.show()
```

## Scatter plot

```python
friends = [70, 65, 72, 63, 71, 64, 60, 64, 67]
minutes = [175, 170, 205, 120, 220, 130, 105, 145, 190]
labels = list('abcdefghi')

plt.scatter(friends, minutes)

# rotule cada ponto
for label, friend_count, minute_count in zip(labels, friends, minutes):
    plt.annotate(
        label,
        xy=(friend_count, minute_count), # posição do rótulo no respectivo ponto
        xytext=(5, -5), # deslocamento -> levemente deslocado
        textcoords='offset points'
    )

plt.title('Daily Minutes vs. Number of Friends')
plt.xlabel('# of friends')
plt.ylabel('daily minutes spent on the site')

plt.show()
```
